```python
# Limpiar strings
fbref['Player'] = fbref['Player'].astype(str).str.strip()
fbref['Squad'] = fbref['Squad'].astype(str).str.strip()
transf['Jugadores'] = transf['Jugadores'].astype(str).str.strip()
transf['equipo'] = transf['equipo'].astype(str).str.strip()
```

```python
from rapidfuzz import process, fuzz
import pandas as pd

# --- PRIMERA FASE: Match por nombre con umbral alto (≥75) ---
nombres_transf = transf['Jugadores'].tolist()

def fuzzy_match_nombre(nombre, lista_objetivo, umbral=75):
    resultado = process.extractOne(nombre, lista_objetivo, scorer=fuzz.token_sort_ratio)
    if resultado and resultado[1] >= umbral:
        return resultado[0], resultado[1]
    return None, resultado[1] if resultado else 0

fbref[['MatchTransf', 'MatchScore']] = fbref['Player'].apply(
    lambda x: pd.Series(fuzzy_match_nombre(x, nombres_transf, umbral=75))
)

# --- SEGUNDA FASE: para los que no han hecho match (MatchScore < 75) ---
# Crea tuplas con (nombre, equipo) de transf
transf_tuplas = list(zip(transf['Jugadores'], transf['equipo']))

# Matching combinado para los casos dudosos
def fuzzy_match_nombre_y_equipo(nombre_fbref, equipo_fbref, transf_tuplas, umbral_nombre=60, umbral_equipo=60):
    mejores = []
    for nombre_t, equipo_t in transf_tuplas:
        score_n = fuzz.token_sort_ratio(nombre_fbref, nombre_t)
        score_e = fuzz.token_sort_ratio(equipo_fbref, equipo_t)
        if score_n >= umbral_nombre and score_e >= umbral_equipo:
            mejores.append((nombre_t, score_n, score_e))
    if mejores:
        mejores.sort(key=lambda x: (x[1] + x[2]), reverse=True)
        return mejores[0][0], mejores[0][1], mejores[0][2]
    return None, 0, 0

# Aplicar solo a los que no fueron emparejados en la primera fase
fbref_dudosos = fbref[fbref['MatchScore'] < 75].copy()

fbref_dudosos[['MatchTransf_alt', 'ScoreNombre_alt', 'ScoreEquipo_alt']] = fbref_dudosos.apply(
    lambda row: pd.Series(fuzzy_match_nombre_y_equipo(row['Player'], row['Squad'], transf_tuplas)),
    axis=1
)

# Rellenar las coincidencias encontradas en esta segunda fase
fbref.loc[fbref['MatchScore'] < 75, 'MatchTransf'] = fbref_dudosos['MatchTransf_alt']
fbref.loc[fbref['MatchScore'] < 75, 'MatchScore'] = fbref_dudosos['ScoreNombre_alt']
fbref.loc[fbref['MatchScore'] < 75, 'ScoreEquipo'] = fbref_dudosos['ScoreEquipo_alt']

# Filtrar jugadores con match
fbref_matcheados = fbref[fbref['MatchTransf'].notna()].copy()

# Merge final con transf
df_merged = pd.merge(
    fbref_matcheados,
    transf,
    left_on='MatchTransf',
    right_on='Jugadores',
    how='left'
)

df_merged.drop(columns=['Jugadores'], inplace=True)
print(f"✅ Total emparejados: {df_merged.shape[0]} jugadores")

# (Opcional) ver los peores matches
df_merged.sort_values(by='MatchScore').head(5)
```

```python
# QUITAMOS DUPLICADOS: mismo MatchTransf emparejado a varios jugadores fbref
# Nos quedamos con el de mayor MatchScore
fbref_final = fbref[fbref['MatchTransf'].notna()].copy()

# Ordenar para que el mejor match quede primero
fbref_final = fbref_final.sort_values(by='MatchScore', ascending=False)

# Eliminar duplicados en MatchTransf (conservando el mejor match)
fbref_final = fbref_final.drop_duplicates(subset='MatchTransf', keep='first')
```

```python
df_merged = pd.merge(
    fbref_final,
    transf,
    left_on='MatchTransf',
    right_on='Jugadores',
    how='left'
)
df_merged.drop(columns=['Jugadores'], inplace=True)
print(f"✅ Emparejamientos únicos: {df_merged.shape[0]} jugadores")
```
