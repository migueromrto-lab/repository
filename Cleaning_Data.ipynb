```python
import pandas as pd
import glob
import os
from functools import reduce

# Ruta a tu carpeta con excels en Drive
ruta = "/content/drive/MyDrive/DATOS FBREF 5 GRANDES LIGAS"

# Obtener lista de archivos .xlsx
archivos = glob.glob(os.path.join(ruta, "*.xlsx"))
print("Archivos encontrados:", archivos)

```

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
dataframes = []

# Establece columnas clave comunes a todos los excels
columnas_clave = ['Jugador', 'País', 'Posc', 'Equipo', 'Comp', 'Nacimiento']

for archivo in archivos:
    df = pd.read_excel(archivo)

    # Eliminar columnas duplicadas (por ejemplo 'Rk' si existe)
    if 'Rk' in df.columns:
        df = df.drop(columns=['Rk'])

    # Asegurarse de que las columnas clave están
    columnas_presentes = [col for col in columnas_clave if col in df.columns]
    df = df.set_index(columnas_presentes)

    # Añadir al listado
    dataframes.append(df)

# Unir horizontalmente por las claves comunes
df_final = pd.concat(dataframes, axis=1).reset_index()

# Eliminar columnas duplicadas si aparecen
df_final = df_final.loc[:, ~df_final.columns.duplicated()]


```

```python
#Función para limpiar duplicados y eliminar jugadores sin participación

def Limpiar_duplicados(df):
    df = df.copy()

    # Ordenar por jugador y 90s (de mayor a menor)
    df = df.sort_values(by=["Jugador", "90 s"], ascending=[True, False])

    # Contar ocurrencias acumuladas de cada jugador
    df['duplicado_n'] = df.groupby('Jugador').cumcount()

    # Añadir sufijo solo si es una ocurrencia duplicada (i.e., n > 0)
    df.loc[df['duplicado_n'] > 0, 'Jugador'] = df['Jugador'] + '_' + (df['duplicado_n'] + 1).astype(str)

    # Eliminar la columna auxiliar
    df.drop(columns='duplicado_n', inplace=True)

    # Limpiar espacios y convertir a numérico
    df['90 s'] = df['90 s'].astype(str).str.strip()
    df['90 s'] = pd.to_numeric(df['90 s'], errors='coerce')

    # Eliminar jugadores sin participación
    df = df[df['90 s'] != 0.0]

    # Reordenar por índice
    df = df.sort_index()

    return df
```

```python
df = Limpiar_duplicados(df)
```

```python
#Función para limpiar errores típicos de FBREF

def Limpiar_errores_fbref(df):
    df = df.copy()

    import pandas as pd
    import numpy as np

    df = df.copy()

    # Eliminar columna 'Rk' si existe
    if 'Rk' in df.columns:
        df = df.drop(columns=['Rk'])

    # Eliminar filas completamente vacías
    df = df.dropna(how='all')

    # Quitar espacios en los nombres de columnas
    df.columns = df.columns.str.strip()

    # Limpiar 'Edad': dejar solo el número de años
    if 'Edad' in df.columns:
        df['Edad'] = df['Edad'].astype(str).str.extract(r'(\d+)').astype(int)

    # Solucionar el problema del país
    df['País'] = df['País'].str.split(' ').str[1]

    #Solucionar el problema de la doble posición
    df['Posc'] = df['Posc'].str.split(',').str[0]

    # Eliminar la primera parte de la columna Comp (liga), y quedarse con la segunda parte
    df['Comp'] = df['Comp'].str.split(' ', n=1).str[1].fillna(df['Comp'])

    # Eliminar filas sin nombre de jugador
    if 'Jugador' in df.columns:
        df = df[df['Jugador'].notna()]

    # Convertir a float las columnas numéricas mal tipadas (si alguna es string)
    for col in df.columns:
        if df[col].dtype == object:
            try:
                df[col] = df[col].str.replace('%', '')  # quitar % si hay
                df[col] = df[col].str.replace(',', '')  # quitar separador de miles
                df[col] = df[col].astype(float)
            except:
                pass  # si no se puede convertir, se deja como está

    #Rellenar los posibles Nan que aparezcan
    df.fillna(0.0, inplace=True)

    return df
```
